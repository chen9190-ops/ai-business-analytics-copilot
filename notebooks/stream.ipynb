```python
%%writefile app.py

import streamlit as st
import pandas as pd

# ==============================
# Import Your Modules
# ==============================

from data_cleaning import clean_data

from ai_agent import answer_business_question

# ==============================
# Session Memory
# ==============================

if "router_history" not in st.session_state:

    st.session_state.router_history = []

if "insight_history" not in st.session_state:

    st.session_state.insight_history = []

# ==============================
# Title
# ==============================

st.title(
    "AI Business Analytics Copilot"
)

st.write(
    "Upload a CSV file and ask business questions."
)

# ==============================
# Upload CSV
# ==============================

uploaded_file = st.file_uploader(
    "Upload CSV File",
    type=["csv"]
)

# ==============================
# Run System
# ==============================

if uploaded_file is not None:

    # ==========================
    # Load Dataset
    # ==========================

    df = pd.read_csv(
        uploaded_file,
        encoding="latin1"
    )

    st.success(
        "Dataset uploaded successfully."
    )

    # ==========================
    # Clean Data
    # ==========================

    with st.spinner(
        "Cleaning data..."
    ):

        df = clean_data(df)

    st.success(
        "Data cleaning completed."
    )

    # ==========================
    # Show Dataset
    # ==========================

    st.subheader(
        "Dataset Preview"
    )

    st.dataframe(
        df.head()
    )

    # ==========================
    # User Question
    # ==========================

    question = st.text_input(
        "Ask a business question"
    )

    # ==========================
    # Analyze Button
    # ==========================

    if st.button("Analyze"):

        with st.spinner(
            "Generating AI insight..."
        ):

            # ======================
            # AI Agent
            # ======================

            answer = answer_business_question(

                question,

                df,

                st.session_state.router_history,

                st.session_state.insight_history
            )

        # ==========================
        # Output
        # ==========================

        st.subheader(
            "AI Business Insight"
        )

        st.write(answer)
```
